# dua

```python
class Car:
    def __init__(self, satu, dua, tiga, empat):
        self.make = satu
        self.model = dua
        self.year = tiga
        self.color = empat
        self.speed = 0

    def accelerate(self, increment=5):
        """Increase the car's speed."""
        self.speed += increment
        return f"The car has accelerated to {self.speed} mph."

    def brake(self, decrement=5):
        """Decrease the car's speed."""
        if self.speed - decrement >= 0:
            self.speed -= decrement
            return f"The car has slowed down to {self.speed} mph."
        else:
            self.speed = 0
            return "The car has stopped."

    def repaint(self, new_color):
        """Change the car's color."""
        self.color = new_color
        return f"The car's color has been changed to {self.color}."

    def __str__(self):
        return f"{self.year} {self.make} {self.model}, {self.color}, traveling at {self.speed} mph."
```

```python
# Create an instance of the Car class
my_car = Car(satu="Toyota", dua="Corolla", tiga=2020, empat="Blue")

# Use the object's __str__ method to print its details
print(my_car)

# Accelerate the car and print its new state
my_car.accelerate(10)
print(my_car)

# Brake the car and print its new state
my_car.brake(7)
print(my_car)

# Repaint the car and print its new state
my_car.repaint("Red")
print(my_car)
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def __str__(self):
        return f"{self.year} {self.make} {self.model}"

class Car_2(Vehicle):
    def __init__(self, make, model, year, color):
        super().__init__(make, model, year)  # Call the __init__ of the Vehicle class
        self.color = color
        self.speed = 0

    def accelerate(self, increment=5):
        """Increase the car's speed."""
        self.speed += increment
        return f"The car has accelerated to {self.speed} mph."

    def brake(self, decrement=5):
        """Decrease the car's speed."""
        if self.speed - decrement >= 0:
            self.speed -= decrement
            return f"The car has slowed down to {self.speed} mph."
        else:
            self.speed = 0
            return "The car has stopped."

    def repaint(self, new_color):
        """Change the car's color."""
        self.color = new_color
        return f"The car's color has been changed to {self.color}."

    def __str__(self):
        vehicle_str = super().__str__()  # Get the string representation from the Vehicle class
        return f"{vehicle_str}, {self.color}, traveling at {self.speed} mph."
        #return f"{self.year} {self.make} {self.model}, {self.color}, traveling at {self.speed} mph."
```

```python
# Assuming the Car class definition from earlier

# Create an instance of the Car class
my_car = Car_2(make="Toyota", model="Corolla", year=2020, color="Blue")

# Accelerate the car by 10 mph
my_car.accelerate(10)

# Print the current state of the car
print(my_car)
```
