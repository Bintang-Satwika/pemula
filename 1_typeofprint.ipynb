```python
# Example of a list
my_list = [1, 2, 3]
my_list.append(4) 
print("List after append:", my_list)
# Example of a tuple
my_tuple = (1, 2, 3)
# Tuples are immutable, attempting to modify a tuple will result in an error
print("Tuple:", my_tuple)
```

```python
my_set = {1, 2, 3, 3, 4}
print("Set:", my_set)
my_set.add("d")
print(my_set)
# List to set
lst = [1, 2, 3, 1]
st = set(lst)
print(st)  # Output: {1, 2, 3}

# Tuple to set
tpl = (1, 2, 3, 2)
st = set(tpl)
print(st)  # Output: {1, 2, 3}

# String to set
string = 'hello'
st = set(string)
print(st)  # Output: {'h', 'e', 'l', 'o'}
```

```python
# Empty dictionary
my_dict = {}

# Dictionary with initial values
my_dict = {'name': 'Alice', 'age': 25, 'city': 'New York'}
print(my_dict)
print(my_dict.keys())
print(my_dict.values())
```

```python
# Accessing values by keys
print(my_dict['name'])  # Output: Alice

# Using get() method
print(my_dict.get('age'))  # Output: 25
print(my_dict.get('country', 'USA'))  # Output: USA (default value if key not found)
```

```python
# Adding a new key-value pair
my_dict['email'] = 'alice@example.com'

# Updating an existing key-value pair
my_dict['age'] = 99
print(my_dict)
```

```python
student_grades = {
    'Alice': [85, 90, 78],
    'Bob': [92, 88, 84],
    'Charlie': [87, 91, 79]
}

for student, grades in student_grades.items():
    print(f"{student}'s grades are: {grades}")
```
