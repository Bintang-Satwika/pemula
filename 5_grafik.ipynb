```python
import matplotlib.pyplot as plt
import numpy as np
import datetime

# Generate a range of dates
start_date = datetime.date(2024, 7, 1)
end_date = datetime.date.today()
delta = datetime.timedelta(days=1)
dates = [start_date + i*delta for i in range((end_date - start_date).days + 1)]

# Generate two sets of data
data1 = np.random.randint(0, 100, len(dates))
data2 = np.random.randint(0, 100, len(dates))  # Second set of data

# Create figure and axis
fig, ax = plt.subplots()

# Plotting both sets of data
ax.plot(dates, data1, label='Data 1')  # Add label for legend
ax.plot(dates, data2, label='Data 2')  # Add label for legend

# Formatting the date labels
plt.gcf().autofmt_xdate()  # Auto rotates dates for better readability

# Setting title and labels
ax.set(title='Simple Graph with Datetime X-axis', xlabel='Date', ylabel='Random Value')

# Adding a legend
ax.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Generate a range of dates
dates = np.arange('2022-01-01', '2022-04-11', dtype='datetime64[D]')

# Generate some random data
data = np.random.randn(len(dates)).cumsum()
# Plotting
plt.figure(figsize=(10, 6))
plt.plot(dates, data)

# Formatting the date labels
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
plt.gca().xaxis.set_major_locator(mdates.DayLocator(interval=15))
plt.gcf().autofmt_xdate() # Rotation

plt.title('Time Series Example')
plt.xlabel('Date')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```
