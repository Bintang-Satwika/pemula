```python
bilangan=10.0
pembagi=0
while True:
    try :
        print(bilangan/pembagi)
        print("pembagian sukses")
        break
    except ZeroDivisionError:
        print("Error, bilangan tidak bisa dibagi dengan 0")
        pembagi += 0.0000003

    finally:
        print("eksekusi selesai")
```
