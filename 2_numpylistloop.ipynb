# tiga

```python
import numpy as np

# Create two 2x2 matrices
matrix1 = np.array([[1, 2], [3, 4]])
matrix2 = np.array([[5, 6], [7, 8]])
print(matrix1)
print(matrix2)
```

```python
# Calculate the dot product
dot_product = np.dot(matrix1, matrix2)
print("dot product")
print(dot_product)

matmul_mat = np.matmul(matrix1, matrix2)
print("matmul_mat")
print(matmul_mat)

scalar_mul= np.multiply(matrix1, matrix2)
print("scalar_mul")
print(scalar_mul)
```

```python
stack_matrix= np.stack((matrix1, matrix2), axis=1)
print(stack_matrix.shape)
print(stack_matrix)
print("\nreshape: ")
print(stack_matrix.reshape(-1,2))
```

```python
vstack_matrix= np.vstack((matrix1, matrix2))
print(vstack_matrix.shape)
print(vstack_matrix)
```

```python
conc_matrix= np.concatenate((matrix1, matrix2), axis=0)
print(conc_matrix.shape)
print(conc_matrix)
```

# empat

```python
a= np.array([-2, -1, 0, 1, 2, 3])
hasil= np.where(a>0, a, 
         np.where(a<0, a*10, 999))
print(hasil)
```

```python
hasil=[]
for i in a:
    print(i)
    if i>0:
        hasil.append(i)
    elif i<0:
        hasil.append(i*10)
    else:    
        hasil.append(999)
print(hasil)
```

```python
hasil = [i if i > 0 else i*10 if i < 0 else 999 for i in a]
print(hasil)
```
